Compute the following probabilities with your justification.

1. P(Alarm = t)

YOUR ANSWER HERE:
$$\begin{aligned} P(A = t) &= P(A = t \mid T = t, F = t)P(T = t)P(F = t) \\ &\quad + P(A = t \mid T = t, F = f)P(T = t)P(F = f) \\ &\quad + P(A = t \mid T = f, F = t)P(T = f)P(F = t) \\ &\quad + P(A = t \mid T = f, F = f)P(T = f)P(F = f) \\ &= (0.5)(0.02)(0.01) + (0.85)(0.02)(0.99) + (0.99)(0.98)(0.01) + (0.0001)(0.98)(0.99) \\ &= 0.0001 + 0.01683 + 0.009702 + 0.00009702 \\ &= 0.02672902 \approx 0.0267 \end{aligned}$$

2. P(Leaving = t)

$$\begin{aligned} P(L = t) &= P(L = t \mid A = t)P(A = t) + P(L = t \mid A = f)P(A = f) \\ &= (0.88)(0.02672902) + (0.001)(0.97327098) \\ &= 0.0235215376 + 0.00097327098 \\ &= 0.02449480858 \approx 0.0245 \end{aligned}$$

5. P(Tampering = t | Alarm = t, Fire = f)

YOUR ANSWER HERE:
$$\begin{aligned} P(T = t \mid A = t, F = f) &= \frac{P(A = t \mid T = t, F = f)P(T = t)}{P(A = t \mid T = t, F = f)P(T = t) + P(A = t \mid T = f, F = f)P(T = f)} \\ &= \frac{(0.85)(0.02)}{(0.85)(0.02) + (0.0001)(0.98)} \\ &= \frac{0.017}{0.017 + 0.000098} \\ &= \frac{0.017}{0.017098} \approx 0.9943 \end{aligned}$$
