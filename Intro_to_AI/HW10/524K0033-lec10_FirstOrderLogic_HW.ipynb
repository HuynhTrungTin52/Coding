## TODO 2:

Translate these English sentences into First-Order Logic


1.   Every student who studies hard will pass every exam.
2.   There exists a teacher who teaches every subject to some student.
3.   Every person who loves someone will eventually be loved by someone.
4.   No country that has a nuclear weapon is trusted by all other countries.
5.   Every employee who is managed by someone who is also an employee gets a bonus if the company makes a profit.
6.   No scientist publishes a paper unless it has been reviewed by at least two other scientists.





$$\begin{aligned}
\text{1.}\;& \forall x \quad (Student(x) \wedge StudiesHard(x)) \Rightarrow \forall e \; (Exam(e) \Rightarrow Passes(x,e)) \\
\text{2.}\;& \exists t, s \quad Teacher(t) \wedge Student(s) \wedge \forall sub \; (Subject(sub) \Rightarrow Teaches(t,sub,s)) \\
\text{3.}\;& \forall x \quad (Person(x) \wedge \exists y \; (Person(y) \wedge Loves(x,y))) \Rightarrow \exists z \; (Person(z) \wedge EventuallyLoves(z,x)) \\
\text{4.}\;& \forall c \quad (Country(c) \wedge \exists w \; (NuclearWeapon(w) \wedge Has(c,w))) \Rightarrow \neg \forall o \; ((Country(o) \wedge o \ne c) \Rightarrow Trusted(o,c)) \\
\text{5.}\;& \forall e,m,co \quad (Employee(e) \wedge Employee(m) \wedge Manages(m,e) \wedge WorksFor(e,co) \wedge Company(co) \wedge MakesProfit(co)) \Rightarrow GetsBonus(e) \\
\text{6.}\;& \forall s,p \quad (Scientist(s) \wedge Publishes(s,p)) \Rightarrow \exists s_1,s_2 \; (Scientist(s_1) \wedge Scientist(s_2) \wedge s_1 \ne s_2 \wedge s_1 \ne s \wedge s_2 \ne s \wedge Reviews(s_1,p) \wedge Reviews(s_2,p))
\end{aligned}$$
