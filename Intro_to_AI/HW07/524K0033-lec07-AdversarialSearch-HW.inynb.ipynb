## MINIMAX Search

Complete TODOs 4 - 7 to implement the MINIMAX search algorithm, in which update nodes' values in H2.

```python
def minimax_search(state=(0, 0), children=H1, values=H2):
  '''
    # TODO 4
    Return the optimal action for MAX, the action is represented as edge,
    i.e. (0, 0)-(1, 0)
  '''
  if terminal_test(state, children, values):
    return None
  best_action = None
  best_value = float("-inf")
  for action in children[state]:
    value = max_value(action, children, values)
    if value > best_value:
      best_value = value
      best_action = (state, action)
  values[state] = best_value
  return best_action

def max_value(state, children=H1, values=H2):
  '''
    # TODO 5
    Return the max value among children's ones.
  '''
  if terminal_test(state, children, values):
    return values[state]
  values[state] = max(min_value(child, children, values) for child in children[state])
  return values[state]

def min_value(state, children=H1, values=H2):
  '''
    # TODO 6
    Return the min value among children's ones.
  '''
  if terminal_test(state, children, values):
    return values[state]
  values[state] = min(max_value(child, children, values) for child in children[state])
  return values[state]

def terminal_test(state, children=H1, values=H2):
  '''
    # TODO 7
    Verify the given state is terminal.
  '''
  return state not in children or len(children.get(state, [])) == 0
```
