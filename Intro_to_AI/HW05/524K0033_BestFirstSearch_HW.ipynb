# Graph class

```python
# Directed, weighted graphs
class Graph:
  def __init__(self):
    self.AL = dict() # adjacency list
    self.V = 0
    self.E = 0
    self.H = dict()

  def __str__(self):
    res = 'V: %d, E: %d\n'%(self.V, self.E)
    for u, neighbors in self.AL.items():
      line = '%d: %s\n'%(u, str(neighbors))
      res += line
    for u, h in self.H.items():
      line = 'h(%d) = %d\n'%(u, h)
      res+=line
    return res

  def print(self):
    print(str(self))

  def load_from_file(self, filename):
    '''
        Example input file:
            V E
            u v w
            u v w
            u v w
            ...
            u1 h1
            u2 h2
            u3 h3
            ...

        # input.txt
        7 8
        0 1 5
        0 2 6
        1 3 12
        1 4 9
        2 5 5
        3 5 8
        3 6 7
        4 6 4
        0 14
        1 13
        2 12
        3 11
        4 10
        5 9
        6 0
    '''
    if os.path.exists(filename):
      with open(filename) as g:
        self.V, self.E = [int(it) for it in g.readline().split()]
        for i in range(self.E):
          line = g.readline()
          u, v, w = [int(it) for it in line.strip().split()]
          if u not in self.AL:
            self.AL[u] = []
          self.AL[u].append((v, w))
        for i in range(self.V):
          line = g.readline()
          u, h = [int(it) for it in line.strip().split()]
          self.H[u] = h
```

```python
class GBFS(BestSearchStrategy):
  def search(self, g: Graph, src: int, dst: int) -> tuple:
    expanded = [] # list of expanded vertices in the traversal order
    path = [] # path from src to dst

    # TODO 1

    if src == dst:
      return expanded, [src]
      
    frontier = [(g.H.get(src), src)] 
    # print(frontier)
    # frontier_costs = {src: 0} 
    explored = set() 
    parent = {src: None}

    while frontier:
      cost,node = heapq.heappop(frontier) 
      # frontier_set.remove(node)
      if node in explored:
        continue
      expanded.append(node)
      explored.add(node)
      
      if node == dst:
        curr = node 
        while curr is not None:
          path.append(curr)
          curr= parent[curr]
        path.reverse()
        return expanded, path
      # print(f"At node {node} : {g.AL.get(node,[])}")
      for v, w in g.AL.get(node, []):
        #BFS doesn't use weight, unpack it just in case
        if v not in explored:
          if v not in parent:
            parent[v] = node
          heapq.heappush(frontier,(g.H.get(v),v))
        #   frontier.append(v)
        #   frontier_set.add(v)
    return expanded, path
```

```python
class AStar(BestSearchStrategy):
  def search(self, g: Graph, src: int, dst: int) -> tuple:
    expanded = [] # list of expanded vertices in the traversal order
    path = [] # path from src to dst

    # TODO 2
    g_n = {src: 0}
    f_n_start= g_n[src]+g.H.get(src) 
    frontier = [(f_n_start, src)] 
    explored = set() 
    parent = {src: None}
    
    while frontier:
      
      f_n, node = heapq.heappop(frontier)
      
      if node in explored:
          continue
      
      
      
        
      expanded.append(node)
      explored.add(node)
      if node == dst:
              curr = node
              while curr is not None:
                path.append(curr)
                curr = parent[curr]
              path.reverse()
              return expanded, path
      for v, w in g.AL.get(node, []):
        new_g_n = g_n[node] + w
        
        if v not in explored and (v not in g_n or new_g_n < g_n[v]):
          g_n[v] = new_g_n
          parent[v] = node
          
          new_f_n = new_g_n + g.H.get(v)
          heapq.heappush(frontier, (new_f_n,v))
        
    return expanded, path
```
