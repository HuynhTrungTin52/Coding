# Graph class

```python
# Directed, weighted graphs
class Graph:
  def __init__(self):
    self.AL = dict() # adjacency list
    self.V = 0
    self.E = 0

  def __str__(self): # toString()
    res = 'V: %d, E: %d\n'%(self.V, self.E)
    for u, neighbors in self.AL.items():
      line = '%d: %s\n'%(u, str(neighbors))
      res += line
    return res

  def print(self):
    print(str(self))

  def load_from_file(self, filename):
    '''
        Example input file:
            V E
            u v w
            u v w
            u v w
            ...

        # input.txt
        7 8
        0 1 5
        0 2 6
        1 3 12
        1 4 9
        2 5 5
        3 5 8
        3 6 7
        4 6 4
    '''
    if os.path.exists(filename):
      with open(filename) as g:
        self.V, self.E = [int(it) for it in g.readline().split()]
        for line in g:
          u, v, w = [int(it) for it in line.strip().split()]
          if u not in self.AL:
            self.AL[u] = []
          self.AL[u].append((v, w))
```

# Search Strategies

```python
class SearchStrategy:
  def search(self, g: Graph, src: int, dst: int) -> tuple:
    expanded = [] # list of expanded vertices in the traversal order
    path = [] # path from src to dst
    return expanded, path
```

```python
class BFS(SearchStrategy):
  def search(self, g: Graph, src: int, dst: int) -> tuple:
    expanded = [] # list of expanded vertices in the traversal order
    path = [] # path from src to dst

    # TODO 1
    if src == dst:
      return expanded, [src]
      
    frontier = [src]  #fifo
    frontier_set = {src} 
    explored = set()
    parent = {src: None}
    
    while frontier:
      node = frontier.pop(0) 
      frontier_set.remove(node)
      
      expanded.append(node)
      explored.add(node)
      # print(f"At node {node} : {g.AL.get(node,[])}")
      for v, w in g.AL.get(node, []):
        #BFS doesn't use weight, unpack it just in case
        if v not in explored and v not in frontier_set:
          parent[v] = node
          
          if v == dst:
            curr = v
            while curr is not None:
              path.append(curr)
              curr = parent[curr]
            path.reverse()
            return expanded, path
            
          frontier.append(v)
          frontier_set.add(v)
            
    return expanded, path #if empty then = failure, i think
```

```python
class DFS(SearchStrategy):
  def search(self, g: Graph, src: int, dst: int) -> tuple:
    expanded = [] # list of expanded vertices in the traversal order
    path = [] # path from src to dst

    # TODO 2
    frontier = [src] # lifo
    explored = set()  
    parent = {src: None}
    
    while frontier:
      node = frontier.pop() 
      
      if node not in explored:
        expanded.append(node)
        explored.add(node) 
        
        if node == dst:
          curr = node
          while curr is not None:
            path.append(curr)
            curr = parent[curr]
          path.reverse()
          return expanded, path
          
        for v, w in reversed(g.AL.get(node, [])):
          if v not in explored and v not in frontier:
            parent[v] = node
            frontier.append(v)
    return expanded, path
```

```python
class UCS(SearchStrategy):
  def search(self, g: Graph, src: int, dst: int) -> tuple:
    expanded = [] # list of expanded vertices in the traversal order
    path = [] # path from src to dst

    # TODO 3
    frontier = [(0, src)] 
    frontier_costs = {src: 0} 
    explored = set() 
    parent = {src: None}
    
    while frontier:
      
      cost, node = heapq.heappop(frontier)
      
      if node not in frontier_costs or frontier_costs[node] < cost:
        #print("1")
          continue
      del frontier_costs[node]
      
      if node == dst:
        curr = node
        while curr is not None:
          path.append(curr)
          curr = parent[curr]
        path.reverse()
        return expanded, path
      
        
      expanded.append(node)
      explored.add(node)
      
      for v, w in g.AL.get(node, []):
        child_cost = cost + w
        
        if v not in explored and v not in frontier_costs:
          parent[v] = node
          frontier_costs[v] = child_cost
          heapq.heappush(frontier, (child_cost, v))
          
        elif v in frontier_costs and child_cost < frontier_costs[v]:
          parent[v] = node
          frontier_costs[v] = child_cost
          heapq.heappush(frontier, (child_cost, v))
    return expanded, path
  
```

```python
class DLS(SearchStrategy):
  def __init__(self, LIM: int):
    self.LIM = LIM

  def search(self, g: Graph, src: int, dst: int) -> tuple:
    expanded = [] # list of expanded vertices in the traversal order
    path = [] # path from src to dst
    def recursive_dls(node, limit):
      if node not in expanded:
          expanded.append(node)
          
      if node == dst:
        return [node] 
        
      elif limit == 0:
        return "cutoff"
        
      else:
        cutoff_occurred = False
        
        for v, w in g.AL.get(node, []):
          result = recursive_dls(v, limit - 1)
          
          if result == "cutoff":
            cutoff_occurred = True
            
          elif result != "failure":
            return [node] + result 
            
        if cutoff_occurred:
          return "cutoff"
        else:
          return "failure"
          
    res = recursive_dls(src, self.LIM)
    path = res if isinstance(res, list) else []
    # TODO 4

    return expanded, path
```

```python
class IDS(SearchStrategy):
  def __init__(self, MAX_LIM: int):
    self.MAX_LIM = MAX_LIM

  def search(self, g: Graph, src: int, dst: int) -> tuple:
    expanded = [] # list of expanded vertices in the traversal order
    # path = [] # path from src to dst
    
    # TODO 5
    for depth in range(self.MAX_LIM + 1):
        
      dls = DLS(depth)
      exp, res_path = dls.search(g, src, dst)
      expanded.extend(exp)
      
      if res_path: 
         return expanded, res_path
         
    return expanded, []
    # return expanded, path
```
